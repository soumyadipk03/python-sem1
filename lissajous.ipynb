```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# GLOBAL VARIABLES — CHANGE ONLY THESE
# ============================================================

m = 2                  # Frequency in Y
n = 1                  # Frequency in X

# Amplitude constraint:
# ">"  -> A1 > A2
# "="  -> A1 = A2
# "<"  -> A1 < A2
AMPLITUDE_RELATION = ">"

# Ratio between amplitudes
# Example:
# 2 means the larger amplitude is twice the smaller one
AMPLITUDE_RATIO = 2

# Phase difference:
# 0 = 0
# 1 = π/4
# 2 = π/2
# 3 = 3π/4
# 4 = π
# ...
# 8 = 2π
DELTA_STEP = 1


# ============================================================
# SET A1 AND A2 ACCORDING TO THE CONSTRAINT
# ============================================================

if AMPLITUDE_RELATION == "=":
    A1 = 1
    A2 = 1

elif AMPLITUDE_RELATION == ">":
    A1 = AMPLITUDE_RATIO
    A2 = 1

elif AMPLITUDE_RELATION == "<":
    A1 = 1
    A2 = AMPLITUDE_RATIO

else:
    raise ValueError('AMPLITUDE_RELATION must be ">", "=" or "<"')


# ============================================================
# PHASE DIFFERENCE
# ============================================================

delta = DELTA_STEP * np.pi / 4


# ============================================================
# PARAMETRIC EQUATIONS
# ============================================================

# Common period:
# If m and n have a common factor, this automatically
# produces the complete figure over 0 to 2π.
t = np.linspace(0, 2*np.pi, 20000)

X = A1 * np.cos(n*t + delta)
Y = A2 * np.cos(m*t)


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(8, 8))

plt.plot(X, Y, linewidth=2)

# Coordinate axes
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

# Keep the correct geometric proportions
plt.xlim(-A1 * 1.1, A1 * 1.1)
plt.ylim(-A2 * 1.1, A2 * 1.1)

plt.xlabel("X")
plt.ylabel("Y")

# Phase displayed properly
if DELTA_STEP == 0:
    delta_text = "0"
elif DELTA_STEP == 1:
    delta_text = "π/4"
elif DELTA_STEP == 2:
    delta_text = "π/2"
elif DELTA_STEP == 3:
    delta_text = "3π/4"
elif DELTA_STEP == 4:
    delta_text = "π"
elif DELTA_STEP == 5:
    delta_text = "5π/4"
elif DELTA_STEP == 6:
    delta_text = "3π/2"
elif DELTA_STEP == 7:
    delta_text = "7π/4"
elif DELTA_STEP == 8:
    delta_text = "2π"
else:
    delta_text = f"{DELTA_STEP}π/4"

plt.title(
    f"Lissajous Figure\n"
    f"m = {m}, n = {n},  "
    f"A₁ {AMPLITUDE_RELATION} A₂,  "
    f"δ = {delta_text}"
)

plt.grid(True, alpha=0.25)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# GLOBAL VARIABLES — CHANGE ONLY THESE
# ============================================================

m = 2                  # Frequency in Y
n = 1                  # Frequency in X

# Amplitude constraint:
# ">"  -> A1 > A2
# "="  -> A1 = A2
# "<"  -> A1 < A2
AMPLITUDE_RELATION = ">"

# Ratio between amplitudes
# Example:
# 2 means the larger amplitude is twice the smaller one
AMPLITUDE_RATIO = 2

# Phase difference:
# 0 = 0
# 1 = π/4
# 2 = π/2
# 3 = 3π/4
# 4 = π
# ...
# 8 = 2π
DELTA_STEP = 1


# ============================================================
# SET A1 AND A2 ACCORDING TO THE CONSTRAINT
# ============================================================

if AMPLITUDE_RELATION == "=":
    A1 = 1
    A2 = 1

elif AMPLITUDE_RELATION == ">":
    A1 = AMPLITUDE_RATIO
    A2 = 1

elif AMPLITUDE_RELATION == "<":
    A1 = 1
    A2 = AMPLITUDE_RATIO

else:
    raise ValueError('AMPLITUDE_RELATION must be ">", "=" or "<"')


# ============================================================
# PHASE DIFFERENCE
# ============================================================

delta = DELTA_STEP * np.pi / 4


# ============================================================
# PARAMETRIC EQUATIONS
# ============================================================

# Common period:
# If m and n have a common factor, this automatically
# produces the complete figure over 0 to 2π.
t = np.linspace(0, 2*np.pi, 20000)

X = A1 * np.cos(n*t + delta)
Y = A2 * np.cos(m*t)


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(8, 8))

plt.plot(X, Y, linewidth=2)

# Coordinate axes
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

# Keep the correct geometric proportions
plt.xlim(-A1 * 1.1, A1 * 1.1)
plt.ylim(-A2 * 1.1, A2 * 1.1)

plt.xlabel("X")
plt.ylabel("Y")

# Phase displayed properly
if DELTA_STEP == 0:
    delta_text = "0"
elif DELTA_STEP == 1:
    delta_text = "π/4"
elif DELTA_STEP == 2:
    delta_text = "π/2"
elif DELTA_STEP == 3:
    delta_text = "3π/4"
elif DELTA_STEP == 4:
    delta_text = "π"
elif DELTA_STEP == 5:
    delta_text = "5π/4"
elif DELTA_STEP == 6:
    delta_text = "3π/2"
elif DELTA_STEP == 7:
    delta_text = "7π/4"
elif DELTA_STEP == 8:
    delta_text = "2π"
else:
    delta_text = f"{DELTA_STEP}π/4"

plt.title(
    f"Lissajous Figure\n"
    f"m = {m}, n = {n},  "
    f"A₁ {AMPLITUDE_RELATION} A₂,  "
    f"δ = {delta_text}"
)

plt.grid(True, alpha=0.25)# ============================================================
# DIRECTION OF MOTION — ANTI-CLOCKWISE
# ============================================================

# Pick a few points along the curve where arrows will be shown
arrow_indices = np.linspace(0, len(t) - 1, 12, dtype=int)

for i in arrow_indices:
    # Tangent vector = direction of motion
    dx = -A1 * n * np.sin(n * t[i] + delta)
    dy = -A2 * m * np.sin(m * t[i])

    # Normalize tangent vector so all arrows have similar size
    magnitude = np.sqrt(dx**2 + dy**2)

    if magnitude > 1e-10:
        dx /= magnitude
        dy /= magnitude

        plt.arrow(
            X[i], Y[i],
            dx * A1 * 0.12,
            dy * A2 * 0.12,
            head_width=min(A1, A2) * 0.035,
            head_length=min(A1, A2) * 0.05,
            length_includes_head=True,
            linewidth=1.2
        )
plt.show()
```
