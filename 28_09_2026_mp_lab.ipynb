```python
from re import X
import matplotlib.pyplot as plt
x_values = [1,2,3,4,5,6,7]
y_values = [3,6,10,15,17,20,27]
# Graph
plt.plot(x_values, y_values)
```

```python
# only points
plt.plot(x_values, y_values, 'o')
```

```python
# dashed lines
plt.plot(x_values, y_values, 'o--')
```

```python
# dotted lines
plt.plot(x_values, y_values, 'o:')
```

```python
# normal line
plt.plot(x_values, y_values, 'o-')
```

```python
# more attributes and styling
plt.plot(x_values, y_values, marker='o', color='red', linestyle='--') # 'o' for circle 's' for square '^' for triangle
```

```python
# more definition
plt.plot(x_values, y_values, marker='^', color='red',ms=10, linestyle='--')
```

```python
# alternate linestyle declaration and also superposing two plots
plt.plot(x_values, y_values, marker='s', color='green', linestyle='dashed')
plt.plot(x_values, y_values, color='red', linestyle='--')
plt.show()
```

```python
# labels
plt.plot(x_values, y_values, marker='s', color='green', linestyle='dashed')
plt.xlabel('x_values')
plt.ylabel('y_label')
plt.title('data title')
plt.show()
```

```python
import math as mt
import numpy as np
# squared function
y_labels = np.array([i**2 for i in x_values])
plt.plot(x_values, y_labels, 'o-')
y1_labels = np.array([i*2 for i in x_values])
plt.plot(x_values, y1_values)
```

```python
# subplots using defining axes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (24,8))
ax1.plot(x_values, y_values)
ax1.set_xlabel('x_values')
ax2.plot(x_values, y1_values)
plt.show()
```
